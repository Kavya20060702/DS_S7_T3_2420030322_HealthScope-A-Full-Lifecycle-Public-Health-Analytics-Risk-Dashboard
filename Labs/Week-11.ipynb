```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("Classification and Metric libraries loaded!")
```

```python
print("----- Decision Tree Classifier -----")

# Features: [Study Hours, Attendance Percentage]
X_train = np.array([
    [2, 50], [3, 60], [4, 65], [5, 70],
    [6, 75], [7, 80], [8, 85], [9, 90]
])
# Labels: 0 = Fail, 1 = Pass
y_train = np.array([0, 0, 0, 1, 1, 1, 1, 1])

# Train Decision Tree Model
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# Test Data
X_test = np.array([[3, 55], [5, 72], [8, 88], [4, 60]])
y_test = np.array([0, 1, 1, 0])

# Predict
y_pred_dt = dt_model.predict(X_test)

# Metrics
print("Actual Labels:", y_test)
print("Predicted Labels:", y_pred_dt)
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_dt))
print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Error Rate:", 1 - accuracy_score(y_test, y_pred_dt))
print("Precision:", precision_score(y_test, y_pred_dt))
print("Recall:", recall_score(y_test, y_pred_dt))
print("F1-Score:", f1_score(y_test, y_pred_dt))
```

```python
print("----- K-Nearest Neighbors Classifier (k=3) -----")

# Train KNN Model
knn_model = KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_train, y_train)

# Predict
y_pred_knn = knn_model.predict(X_test)

# Metrics
print("Actual Labels:", y_test)
print("Predicted Labels:", y_pred_knn)
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_knn))
print("Accuracy:", accuracy_score(y_test, y_pred_knn))
print("Error Rate:", 1 - accuracy_score(y_test, y_pred_knn))
print("Precision:", precision_score(y_test, y_pred_knn))
print("Recall:", recall_score(y_test, y_pred_knn))
print("F1-Score:", f1_score(y_test, y_pred_knn))
```
