```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

print("First 5 rows:")
print(df.head())

mean_values = df.mean()

print("\nMean:")
print(mean_values)

median_values = df.median()

print("\nMedian:")
print(median_values)

mode_values = df.mode().iloc[0]

print("\nMode:")
print(mode_values)
```

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

range_values = df.max() - df.min()

print("Range:")
print(range_values)

variance_values = df.var()

print("\nVariance:")
print(variance_values)

std_values = df.std()

print("\nStandard Deviation:")
print(std_values)

Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)

IQR = Q3 - Q1

print("\nInterquartile Range:")
print(IQR)
```

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

skewness_values = df.skew()

print("Skewness:")
print(skewness_values)

kurtosis_values = df.kurt()

print("\nKurtosis:")
print(kurtosis_values)
```
