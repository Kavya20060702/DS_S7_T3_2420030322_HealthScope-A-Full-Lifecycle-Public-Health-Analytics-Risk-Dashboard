```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import make_blobs

print("Clustering libraries loaded!")
```

```python
# -----------------------------
# 1. CREATE SYNTHETIC DATASET
# -----------------------------
X, y = make_blobs(
    n_samples=100,
    centers=3,
    cluster_std=1.5,
    random_state=42
)

# -----------------------------
# 2. K-MEANS CLUSTERING
# -----------------------------
print("----- K-Means Clustering -----")
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

kmeans_labels = kmeans.labels_
centers = kmeans.cluster_centers_

print("Cluster Labels:\n", kmeans_labels)
print("\nCluster Centers:\n", centers)

# Visualization
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=kmeans_labels)
plt.scatter(centers[:, 0], centers[:, 1], marker='X', s=200, c='red')
plt.title("K-Means Clustering")
plt.xlabel("Feature 1 (Annual Income)")
plt.ylabel("Feature 2 (Spending Score)")
plt.show()

# -----------------------------
# 3. HIERARCHICAL CLUSTERING
# -----------------------------
print("----- Hierarchical Clustering -----")
linked = linkage(X, method='ward')

# Visualization
plt.figure(figsize=(10, 5))
dendrogram(linked)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```
