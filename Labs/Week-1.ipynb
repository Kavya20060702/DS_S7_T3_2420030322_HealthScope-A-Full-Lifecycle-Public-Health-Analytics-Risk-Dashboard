```python
#datatypes
x=1  #int
y=2.8  #float
z=1j  #complex
#to verify the type of any object in python, use type() function:
print(x,type(x))
print(y, type(y))
print(z, type(z))
```

```python
x = 1.10
y = 1.0
z = -35.59
print(type(x))
print(type(y))
print(type(z))  
```

```python
x = int(1) #x will be 1
y = int(2.8) #y will be 2
z = int("3") #z will be 3
#in type casting , since we used int the values will be converted as int type only
print(x,y,z)
```

```python
x = float(1)
y = float(2.8)
z = float("3")
w = float("4.2")
print(x,y,z,w)
```

```python
x = str("s1") #x will be 's1'
y = str(2)
z = str("NO")
print(x,y,z)
```
