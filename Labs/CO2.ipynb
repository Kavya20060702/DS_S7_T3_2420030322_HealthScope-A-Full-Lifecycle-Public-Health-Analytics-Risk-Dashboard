```python
#dashboards in python
#improt the main dashboard library
#dash core components, used for interactive components such as graphs, dropdowns, sliders, buttons,etc.
#html create html elements such as headings, paragraphs, divisions,etc.

#creates the new dash application using variable name called app
#__name__ is a special python variable that identifie3s the current python module
import dash
from dash import dcc, html

# Initialize the Dash application
app = dash.Dash(__name__)

# Define the visual layout of the dashboard
app.layout = html.Div([
    html.H1("Dash Tutorial"),
    
    dcc.Graph(
        id="example-graph",
        figure={
            "data": [
                {
                    "x": [1, 2, 3, 4, 5],
                    "y": [5, 4, 7, 4, 8],
                    "type": "line",
                    "name": "Trucks"
                },
                {
                    "x": [1, 2, 3, 4, 5],
                    "y": [6, 3, 5, 3, 7],
                    "type": "bar",
                    "name": "Ships" 
                } 
            ],
            "layout": {
                "title": "Basic Dashboard"
            }
        }
    )
])

# Ensure script runs only when executed directly
if __name__ == "__main__":
    app.run(debug=True, port=8051)
```

```python
#callbacks
import dash
from dash import dcc, html, Input,Output
#create dashboard application
app = dash.Dash(__name__)
#define layout
app.layout=html.Div([
    html.H1("Square Calculator"),
    #input box
    dcc.Input(
        id='input',
        type='text',
        placeholder='Enter a number'
    ),
    #output area
    html.Div(
        id='output'
    )
])
#callback
@app.callback(
    Output(component_id='output',component_property='children'),
    Input(component_id='input',component_property='value')
)
def update_value(input_data):
    try:
        return str(float(input_data) ** 2)
    except:
        return "Error, the input is not a number"
#run application
app.run(debug=True)
```
